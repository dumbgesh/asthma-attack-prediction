```python
#after dropping id and t, we have the following features and target

X = df.drop(columns=["Asthma exacerbation", "id", "t"])
y = df["Asthma exacerbation"]

print("Features:", X.columns.tolist())
print("Feature shape:", X.shape)
```

```python
#check whether Ath has any unusual missing/special values

X["Ath"].value_counts().sort_index()
```

```python
from sklearn.preprocessing import OrdinalEncoder

categorical_features = X.columns.tolist()

encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_encoded = encoder.fit_transform(X[categorical_features])

print("Encoded shape:", X_encoded.shape)
print("Number of features:", X_encoded.shape[1])
```

```python
pd.DataFrame(
    X_encoded,
    columns=X.columns
).head()
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("Classes:", label_encoder.classes_)
print("Encoded target:", np.unique(y_encoded))
```

```python
from sklearn.model_selection import train_test_split

patient_ids = df["id"].unique()

train_ids, test_ids = train_test_split(
    patient_ids,
    test_size=0.20,
    random_state=42
)

print("Total patients:", len(patient_ids))
print("Training patients:", len(train_ids))
print("Testing patients:", len(test_ids))
```

```python
id_overlap = set(train_ids) & set(test_ids)

print("Overlapping patients:", id_overlap)
print("Number of overlapping patients:", len(id_overlap))
```

```python
train_mask = df["id"].isin(train_ids)
test_mask = df["id"].isin(test_ids)

X_train = df.loc[train_mask].drop(columns=["id", "t", "Asthma exacerbation"])
X_test = df.loc[test_mask].drop(columns=["id", "t", "Asthma exacerbation"])

y_train = df.loc[train_mask, "Asthma exacerbation"]
y_test = df.loc[test_mask, "Asthma exacerbation"]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
encoder = OrdinalEncoder(
    handle_unknown="use_encoded_value",
    unknown_value=-1
)

X_train_encoded = encoder.fit_transform(X_train)
X_test_encoded = encoder.transform(X_test)

print("X_train_encoded:", X_train_encoded.shape)
print("X_test_encoded:", X_test_encoded.shape)
```

```python
label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(y_train)
y_test_encoded = label_encoder.transform(y_test)

print("Classes:", label_encoder.classes_)
print("y_train_encoded:", np.unique(y_train_encoded))
print("y_test_encoded:", np.unique(y_test_encoded))
```

```python
print("Minimum training encoded value:", X_train_encoded.min())
print("Minimum testing encoded value:", X_test_encoded.min())
```

```python
import numpy as np

# Number of known categories for each feature
n_categories = np.array([
    len(categories) for categories in encoder.categories_
])

# Give unseen test categories a new valid category index
for i in range(X_test_encoded.shape[1]):
    X_test_encoded[:, i] = np.where(
        X_test_encoded[:, i] == -1,
        n_categories[i],
        X_test_encoded[:, i]
    )

# Tell the model about the additional unknown category
n_categories = n_categories + 1

print("Remaining negative values:", (X_test_encoded < 0).sum())
print("Number of categories per feature:", n_categories)
```
